# Electromagnetic Wave Simulator

This notebook visualizes a simple plane electromagnetic wave in space. The electric field and magnetic field oscillate perpendicular to each other and to the direction of propagation.

For a wave propagating in the +x direction, the fields can be modeled as:

- $E(x,t) = E_0 \sin(kx - \omega t + \phi)$
- $B(x,t) = B_0 \sin(kx - \omega t + \phi)$

where:

- $k$ is the wave number in rad/m
- $\omega$ is the angular frequency in rad/s
- $E_0$ and $B_0$ are amplitudes in V/m
- $\phi$ is the phase shift in radians

Use the sliders below to explore how frequency, wavelength, amplitude, phase, and direction affect the wave.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
import ipywidgets as widgets

try:
    from google.colab import output
    output.enable_custom_widget_manager()
except ImportError:
    pass
```

```python
# Inline EM wave animation for Jupyter and Google Colab
x = np.linspace(0, 4 * np.pi, 600)


def make_wave(amplitude=1.0, frequency=1.0, wavelength=2.0, phase=0.0, direction=1):
    k = 2 * np.pi / wavelength
    omega = 2 * np.pi * frequency
    E = direction * amplitude * np.sin(k * x + phase)
    B = 0.8 * amplitude * np.sin(k * x + phase)
    return E, B, k, omega


def display_wave(amplitude=1.0, frequency=1.0, wavelength=2.0, phase=0.0, direction=1):
    k = 2 * np.pi / wavelength
    omega = 2 * np.pi * frequency

    fig, ax = plt.subplots(figsize=(10, 5))
    ax.set_xlim(0, 4 * np.pi)
    ax.set_ylim(-1.8 * amplitude, 1.8 * amplitude)
    ax.set_xlabel('Distance x (m)')
    ax.set_ylabel('Field amplitude (V/m)')
    ax.set_title(f'Electromagnetic Wave Simulation | Phase: {phase:.2f} rad')
    ax.grid(True, alpha=0.3)

    line_E, = ax.plot([], [], color='tab:blue', lw=2.5, label='Electric field E')
    line_B, = ax.plot([], [], color='tab:orange', lw=2, label='Magnetic field B')
    ax.legend(loc='upper right')

    def update(frame):
        t = frame / 20
        E = direction * amplitude * np.sin(k * x - omega * t + phase)
        B = 0.8 * amplitude * np.sin(k * x - omega * t + phase)
        line_E.set_data(x, E)
        line_B.set_data(x, B)
        return line_E, line_B

    anim = FuncAnimation(fig, update, frames=np.arange(0, 100), blit=True, interval=40)
    animation_html = HTML(anim.to_jshtml())
    plt.close(fig)
    return animation_html


amplitude_slider = widgets.FloatSlider(value=1.0, min=0.2, max=2.0, step=0.1, description='Amplitude (V/m)')
frequency_slider = widgets.FloatSlider(value=1.0, min=0.2, max=3.0, step=0.1, description='Frequency (Hz)')
wavelength_slider = widgets.FloatSlider(value=2.0, min=0.5, max=6.0, step=0.1, description='Wavelength (m)')
phase_slider = widgets.FloatSlider(value=0.0, min=0.0, max=2 * np.pi, step=0.1, description='Phase (rad)')
direction_dropdown = widgets.Dropdown(options=[('Forward (+x)', 1), ('Backward (-x)', -1)], value=1, description='Direction')

wave_controls = widgets.interactive(
    display_wave,
    amplitude=amplitude_slider,
    frequency=frequency_slider,
    wavelength=wavelength_slider,
    phase=phase_slider,
    direction=direction_dropdown,
)

wave_controls
```

```python
# Wave values for quick reference
E0, B0, k, omega = make_wave(amplitude=1.0, frequency=1.0, wavelength=2.0, phase=0.0, direction=1)
print(f'Angular frequency (omega) = {omega:.3f} rad/s')
print(f'Wave number (k) = {k:.3f} rad/m')
print(f'Propagation speed (v = omega / k) = {omega / k:.3f} units/s')
```

## Interpretation

The electric and magnetic fields oscillate in phase with each other, which is a key property of electromagnetic waves. Changing the frequency changes the time variation of the wave, while changing the wavelength changes how spread out the oscillations are across space. The direction control flips the wave propagation along the x-axis.

You can use the sliders to explore how the wave changes in real time.
